```python
a = 1
while a<=9:
    b = 1
    while b<=9:
        print("*",end=' ')
        b += 1
    a += 1
    print()
   
```

```python
row = int(input("Enter a row number : "))
column = int(input("Enter a column number :"))
symbol = input ("Enter a symbol number :")

a = 1
while a <=row:
    b = 1
    while b<=column:
        print(symbol,end=' ')
        b += 1
    a +=1
    print()
```
